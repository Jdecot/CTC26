```python
import pandas as pd
import time
import datetime
filename = 'cronos.csv'
cronos_prices_file = f'D:/Data_crypto_tax_calculator/CROEUR_bitget/cronos.csv'
# cronos_df = pd.read_csv(kraken_data_price, names=["timestamp","price","unknown"], sep=',')
cronos_df = pd.read_csv(cronos_prices_file, sep=';')

cronos_df.head(5)


```

```python
def convert_timestamp_scientist_to_date(time_trade):
    scientist_timestamp = time_trade.replace(',', '.')
    full_timestamp = int(float(scientist_timestamp)) / 1000  # Convertir en float puis en int
    date_standard = datetime.datetime.fromtimestamp(full_timestamp).strftime('%Y-%m-%d %H:%M:%S')
    return date_standard


def convert_row(row_index, cronos_df):
    timeOpen = cronos_df.loc[row_index, 'timeOpen']
    timeClose = cronos_df.loc[row_index, 'timeClose']
    timeHigh = cronos_df.loc[row_index, 'timeHigh']
    timeLow = cronos_df.loc[row_index, 'timeLow']

    priceOpen = float(cronos_df.loc[row_index, 'priceOpen'].replace(',', '.'))
    priceHigh = float(cronos_df.loc[row_index, 'priceHigh'].replace(',', '.'))
    priceLow = float(cronos_df.loc[row_index, 'priceLow'].replace(',', '.'))
    priceClose = float(cronos_df.loc[row_index, 'priceClose'].replace(',', '.'))

    dateOpen = convert_timestamp_scientist_to_date(timeOpen)
    dateClose = convert_timestamp_scientist_to_date(timeClose)
    dateHigh = convert_timestamp_scientist_to_date(timeHigh)
    dateLow = convert_timestamp_scientist_to_date(timeLow)

    row_as_dict = {
        dateOpen : priceOpen,
        dateClose : priceClose,
        dateHigh : priceHigh,
        dateLow	: priceLow
    }
    
    return row_as_dict

print(convert_row(0, cronos_df))
print(convert_row(len(cronos_df)-1, cronos_df))


# compute_average()
#     print(dateOpen)
#     print(dateClose)
#     print(dateHigh)
#     print(dateLow)
#     somme = priceOpen + priceHigh + priceLow + priceClose
#     average = somme/4
#     print(average)

```
